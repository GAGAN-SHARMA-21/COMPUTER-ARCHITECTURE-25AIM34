```python

# Program to perform ROR and RCR operations
# on 8-bit binary numbers

def ror(binary, positions):
    value = int(binary, 2)

    print("\nROTATE RIGHT WITHOUT CARRY (ROR)")
    print("Original Binary:", binary)

    for i in range(positions):
        lsb = value & 1
        value = (value >> 1) | (lsb << 7)

        print("Step", i + 1, ":", format(value, '08b'))

    print("Final ROR Result:", format(value, '08b'))


def rcr(binary, positions, carry):
    value = int(binary, 2)

    print("\nROTATE RIGHT WITH CARRY (RCR)")
    print("Original Binary:", binary)
    print("Initial Carry Flag:", carry)

    for i in range(positions):
        old_lsb = value & 1

        value = (value >> 1) | (carry << 7)
        carry = old_lsb

        print("Step", i + 1, ":",
              format(value, '08b'),
              "Carry:", carry)

    print("Final RCR Result:", format(value, '08b'))
    print("Final Carry Flag:", carry)


# Given binary number
number = "11010101"

# Perform operations
ror(number, 3)
rcr(number, 3, 0)
```
