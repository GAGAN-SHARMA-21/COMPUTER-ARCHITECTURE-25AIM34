```python

# Program to perform signed binary arithmetic
# using 8-bit 2's complement representation

def to_binary(n):
    return format(n & 0xFF, '08b')


def arithmetic(a, b, operation):

    print("\nExpression:", a, operation, b)

    print("First operand :", to_binary(a))
    print("Second operand:", to_binary(b))

    if operation == '+':
        second = b
    else:
        second = -b

    print("Operand used for addition:", to_binary(second))

    result = a + second

    print("\nBinary Addition:")
    print(to_binary(a))
    print("+", to_binary(second))
    print("----------")

    print("Result in binary :", to_binary(result))
    print("Result in decimal:", result)

    if result < -128 or result > 127:
        print("Arithmetic Overflow: YES")
    else:
        print("Arithmetic Overflow: NO")


# Given expressions
arithmetic(46, 29, '+')
arithmetic(-48, -23, '-')
```
